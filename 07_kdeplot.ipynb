## **KDE PLOT**
```text
kdeplot() draws the KDE curve directly, without the histogram.

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

````python
tips.tail(10)
````

### **Plotting Our KDE**

````python
sns.kdeplot(
    data=tips,
    x="total_bill"
)

plt.show()
````

### **hue**

````python
sns.kdeplot(
    data=tips,
    x="total_bill",
    hue="smoker"
)

plt.show()
````

### **fill**

#### *without grouping*

````python
sns.kdeplot(
    data=tips,
    x="total_bill",
    fill=True,
    color="red",
)

plt.show()
````

#### *with grouping*

````python
sns.kdeplot(
    data=tips,
    x="total_bill",
    hue="smoker",
    fill=True
)

plt.show()
````

````python
sns.kdeplot(
    data=tips,
    x="total_bill",
    hue="smoker",
    fill=True,
    palette=["red","green"] # customized color
)

plt.show()
````
