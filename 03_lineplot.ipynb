## **Line Plot**
```text
A line plot is useful for showing how a numerical value changes across an ordered variable, especially over time.

### **Importing seaborn and matplotlib and loading flight dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

print(sns.get_data_home()) # had to check where our datasets are stored 
````

````python
flights = sns.load_dataset("flights")
flights.head()
````

### **Plotting Line Plot**

````python
sns.lineplot(
    data=flights,
    x="year",
    y="passengers"
)

plt.show()
````

````python
sns.lineplot(
    data=flights,
    x="year",
    y="passengers",
    hue="month"
)

plt.show()
````

```text
hue helped to create different lines for each month
