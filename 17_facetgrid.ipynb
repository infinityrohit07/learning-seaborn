## **Facet Grid**
```text
FacetGrid() is used to create multiple plots for different groups of your data.

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

### **Plotting our Facet Grid**
```text
Same type of plot, but split into separate panels based on a categorical variable.

````python
g = sns.FacetGrid(
    tips,
    col="time"
)

g.map(sns.scatterplot, "total_bill", "tip")

plt.show()
````

`col="time"`
```text
Create a separate column/panel for each unique value in time.
```

```text
Since time contains:
1. Lunch
2. Dinner
Seaborn creates two plots.

Then:
```
`g.map(sns.scatterplot, "total_bill", "tip")`
```text
puts the same scatter plot into each panel.


````python
g = sns.FacetGrid(
    tips,
    row="time" # vertically aligns the plots
)

g.map(sns.scatterplot, "total_bill", "tip")

plt.show()
````

````python
g = sns.FacetGrid(
    tips,
    row="time", # vertically split into lunch and dinner
    col="sex" 
)

g.map(sns.scatterplot, "total_bill", "tip")

plt.show()
````
