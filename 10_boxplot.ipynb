### **Box Plot**
```text
A box plot summarizes the distribution of numerical data using quartiles, median, and potential outliers.

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

### **Plotting Box Plot**

````python
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.show()
````

```text
A box plot has several important parts:

          •        ← potential outlier
          |
          |        ← upper whisker
      ┌───────┐
      │       │
      │   ─── │ ← median
      │       │
      └───────┘
          |
          |        ← lower whisker

The box represents the middle 50% of the data:
Bottom of box → Q1 (25th percentile)
Line inside box → Median (50th percentile)
Top of box → Q3 (75th percentile)

```text
We have discussed in detail about box plots in our matplotlib learning sessions. Take references from it to understand our box plot if u still have confusion.

### **hue**

````python
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)

plt.show()
````

### *Common Parameters*

| Parameter   | Purpose                 |
| ----------- | ----------------------- |
| `data`      | DataFrame               |
| `x`         | Categorical variable    |
| `y`         | Numerical variable      |
| `hue`       | Divide data into groups |
| `dodge`     | Separate hue groups     |
| `order`     | Control category order  |
| `hue_order` | Control hue order       |
| `palette`   | Set colors              |
| `width`     | Control box width       |
| `fill`      | Filled or outline boxes |
| `linewidth` | Box outline thickness   |
