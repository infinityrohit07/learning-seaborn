## **Histogram**
```text
A histogram is used to understand the distribution of a numerical variable.

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

````python
tips.head(10)
````

### **Plotting Histogram**

````python
hists = sns.histplot(
    data=tips,
    x="total_bill"
)

plt.bar_label(hists.containers[0])
plt.show()
````

remember it gives range.

### **bins**
```text
bins tells Seaborn to divide the range of total_bill into number of intervals we want.

````python
hists = sns.histplot(
    data=tips,
    x="total_bill",
    bins=5              # bins=5 tells Seaborn to divide the range of total_bill into 5 intervals.
)
plt.bar_label(hists.containers[0])
plt.show()
````

### **hue in Histogram**

````python
hists = sns.histplot(
    data=tips,
    x="total_bill",
    bins=5,
    hue="smoker",
)
plt.bar_label(hists.containers[0])
plt.show()
````

```text
Blue layer  = smokers
Orange layer = non-smokers
Grey = they overlap

**Same just bar labeling both of our category.**

````python
hists = sns.histplot(
    data=tips,
    x="total_bill",
    bins=5,
    hue="smoker"
)

plt.bar_label(
    hists.containers[0],
)

plt.bar_label(
    hists.containers[1],
    label_type="center"
)

plt.show()
````

| `label_type` | Position        |
| ------------ | --------------- |
| `"edge"`     | End/edge of bar |
| `"center"`   | Center of bar   |


### **multiple="dodge"** 
```text
It tells Seaborn not to overlap the groups. Instead, it places them side-by-side within each bin.

````python
hists = sns.histplot(
    data=tips,
    x="total_bill",
    bins=5,
    hue="smoker",
    multiple="dodge",
    palette="colorblind"
)

plt.bar_label(
    hists.containers[0],
)

plt.bar_label(
    hists.containers[1],
)

plt.show()
````

### **stat in Histogram** 

By default:

`sns.histplot(
    data=tips,
    x="total_bill"
)`

uses:

stat="count"

So the y-axis shows:

How many observations are in each bin?

**stat="count"** - How many observations are in each bin?

````python
hists = sns.histplot(
    data=tips,
    x="total_bill",
    stat="count",
    bins=5
)

plt.bar_label(hists.containers[0])
plt.show()
````

**stats = "frequency"** - represents frequency per unit of the x-axis.

````python
hists = sns.histplot(
    data=tips,
    x="total_bill",
    stat="frequency",
    bins=5
)

plt.bar_label(hists.containers[0])

plt.show()
````

**stat="probability"** - represent the proportion of observations.
```text
0.10 → 10% of observations
0.25 → 25% of observations

The total probability across all bins is approximately:
1.0

````python
hists = sns.histplot(
    data=tips,
    x="total_bill",
    stat="probability",
    bins=5
)

plt.bar_label(hists.containers[0])

plt.show()
````

**stats = "percent"**
```text
Same basic idea as probability, but expressed as percentages:

0.10 probability → 10%
0.25 probability → 25%

So the y-axis becomes:
Percent

````python
hists = sns.histplot(
    data=tips,
    x="total_bill",
    stat="percent",
    bins=5
)
plt.bar_label(hists.containers[0], fmt="%1.1f%%")
plt.show()
````

**stats = "density"** 
```text
This converts the histogram into a probability density.

The important concept is that the area under the histogram sums to approximately 1, rather than the bar heights themselves summing to 1.

````python
hists = sns.histplot(
    data=tips,
    x="total_bill",
    stat="density",
    bins=5
)
plt.bar_label(hists.containers[0])
plt.show()
````

### **kde=True**

KDE stands for Kernel Density Estimate.

It adds a smooth curve showing the overall shape of the numerical distribution.

````python
sns.histplot(
    data=tips,
    x="total_bill",
    bins=5,
    kde=True
)

plt.show()
````
