## **Swarm Plot**
```text
A swarm plot is similar to a stripplot(), but Seaborn automatically adjusts the position of points so that they don't overlap.

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

### **Plotting Our First Swarm Plot**

````python
sns.swarmplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.show()
````

```text
Unlike our strip plot  here none of the dots overlap.

### *Common Parameter*
| Parameter   | Purpose                         |
| ----------- | ------------------------------- |
| `data`      | DataFrame                       |
| `x`         | Categorical variable            |
| `y`         | Numerical variable              |
| `hue`       | Divide observations into groups |
| `dodge`     | Separate hue groups             |
| `size`      | Point size                      |
| `alpha`     | Transparency                    |
| `order`     | Category order                  |
| `hue_order` | Hue order                       |
| `palette`   | Colors                          |


````python
sns.swarmplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    dodge=True,
    palette="Set2",
    size=5,
    alpha=0.8
)

plt.title("Total Bill Distribution by Day and Sex")
plt.xlabel("Day")
plt.ylabel("Total Bill")

plt.show()
````
