## **Pair Grid**
```text
pairplot() gives us a ready-made relationship grid, 
while PairGrid() lets us decide what kind of plot goes into different parts of the grid.

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

### **Plotting our Pair Grid**

````python
g = sns.PairGrid(tips)

g.map(sns.scatterplot)

plt.show()
````

### **Why use PairGrid()?**
```text
With pairplot():
```
`sns.pairplot(tips)`
```text
Seaborn decides the structure for us.

With PairGrid():
```
`g = sns.PairGrid(tips)`
`g.map(sns.scatterplot)`
```text
we get access to the grid object g, allowing us to customize different sections of the grid separately.

### **Customizing different section of Pair Grid**

````python
g = sns.PairGrid(tips)

g.map_upper(sns.scatterplot)
g.map_lower(sns.regplot)
g.map_diag(sns.histplot)

plt.show()
````

`map_upper()`
```text
Controls the plots above the diagonal.
In this example, scatter plots are displayed in the upper section.
```


`map_lower()`
```text
Controls the plots below the diagonal.
In this example, regression plots are displayed in the lower section.
```


`map_diag()`
```text
Controls the plots on the diagonal.
In this example, histograms are displayed on the diagonal.

```text
                 Columns
            A        B        C
        ┌────────┬────────┬────────┐
     A  │  DIAG  │ UPPER  │ UPPER  │
        ├────────┼────────┼────────┤
     B  │  LOWER │  DIAG  │ UPPER  │
        ├────────┼────────┼────────┤
     C  │  LOWER │ LOWER  │  DIAG  │
        └────────┴────────┴────────┘

UPPER → map_upper()
LOWER → map_lower()
DIAG  → map_diag()
