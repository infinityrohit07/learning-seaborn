## **Heat Map**
```text
A heatmap represents numerical values using colors.

It is especially useful for:
1. correlation matrices
2. comparing values across rows and columns
3. finding patterns in a matrix
4. quickly identifying high and low values

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

### **Plotting the heat map**

````python
data = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

sns.heatmap(data)

plt.show()
````

### **annot=True**
```text
By default, the heatmap mainly shows the colors.
If we also want the actual numerical values displayed inside each cell:

````python
sns.heatmap(
    data,
    annot=True
)

plt.show()
````

### **cmap**
```text
We can choose the color scheme:

````python
sns.heatmap(
    data,
    annot=True,
    cmap="Blues"
)

plt.show()
````

Some useful colormaps:

### **Correlation Matrix**
```text
A correlation matrix shows the correlation between multiple numerical variables.
In Pandas, we can create one with:

````python
corr = tips.corr(numeric_only=True)

print(corr)
````

**Understanding the values**
```text
Correlation ranges from -1 to +1:

Correlation Meaning:
+1 Perfect positive relationship
0  No linear relationship
-1 Perfect negative relationship

````python
sns.heatmap(
    corr,
    annot=True,         # Show values inside cells
    fmt=".2f",          # Format displayed values
    cmap="coolwarm",    # Color scheme
    linewidths=0.5,     # Space/lines between cells
    cbar=False          # Show/hide color bar
)

plt.title("Correlation Matrix")
plt.show()
````
