### **Importing seaborn and matplotlib and loading tips dataset**

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
```

```python
# Set the size of the figure: width=8 inches, height=5 inches
plt.figure(figsize=(8, 5))

# Create a scatter plot using the tips DataFrame
# x → total bill, y → tip, hue → separate points by sex
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex"
)

# Add a title to the plot
plt.title("Total Bill vs Tip")

# Label the x-axis
plt.xlabel("Total Bill")

# Label the y-axis
plt.ylabel("Tip")

# Save the plot as a PNG image
# dpi=300 → high-resolution image
# bbox_inches="tight" → prevents labels/title from being unnecessarily cut off
plt.savefig(
    "total_bill_vs_tip.png",
    dpi=300,
    bbox_inches="tight"
)

# Display the plot
plt.show()
```
