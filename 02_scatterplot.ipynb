## **Scatter Plot**
```text
A scatter plot displays individual observations as points to show the relationship between two numerical variables.

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

````python
tips
````

### **Plotting Scatter Plot**

````python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
````

### **Customizing Scatter Plot with Matplotlib**

````python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.show()
````

#### **hue**
```text
hue uses the values of a column to divide the data into groups and assigns a different color to each group.

````python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker"
)

plt.show()
````

```text
here the smokers and non-smokers are differentiated easily.

### **style**
```text
style uses a column's values to give different groups different marker shapes.

````python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    style="time"
)

plt.show()
````

### **size**
```text
size uses a column's values to control the size of each marker.

````python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="size",
    size="size"
)

plt.show()
````

```text
we can always combine hue, style and size based on our preferences
