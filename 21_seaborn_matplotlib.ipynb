## **Seaborn + Matplotlib**

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

### *Figure size*
```text
We can control the size using Matplotlib:

````python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
````

### *Title and labels*

````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    errorbar=None
)

plt.title("Average Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Total Bill")

plt.show()
````

### *Axis limits*

````python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.xlim(0, 50)
plt.ylim(0, 12)

plt.show()
````

### *Grid*

````python
sns.set_theme(style="whitegrid")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.grid(axis="y", alpha=0.3)

plt.show()
````

### *Legends*

````python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex"
)

plt.legend(title="Sex")

plt.show()
````

### **Combining them all**

````python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    palette="Set2"
)

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.xlim(0, 55)
plt.ylim(0, 12)

plt.grid(axis="y", alpha=0.3)

plt.legend(title="Sex")

plt.show()
````
