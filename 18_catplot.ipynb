## **Cat Plot**
```text
catplot() is a figure-level function used to create categorical plots.

Instead of using separate functions like:
```
`sns.barplot()`
`sns.countplot()`
`sns.boxplot()`
`sns.violinplot()`
```text
we can use:
```
`sns.catplot()`
```text
and choose the type of categorical plot using kind.

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

### **Plotting our Cat Plot**

````python
sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    kind="bar"
)

plt.show()
````

**Common Options**

| `kind`     | Creates     |
| ---------- | ----------- |
| `"strip"`  | Strip plot  |
| `"swarm"`  | Swarm plot  |
| `"box"`    | Box plot    |
| `"violin"` | Violin plot |
| `"boxen"`  | Boxen plot  |
| `"bar"`    | Bar plot    |
| `"count"`  | Count plot  |
| `"point"`  | Point plot  |


### **Why use catplot()?**
```text
Its biggest advantage becomes clear when we want multiple panels.

````python
sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    col="time",
    row="sex",
    kind="swarm"
)

plt.show()
````

```text
We don't need to memorise the different plot.
