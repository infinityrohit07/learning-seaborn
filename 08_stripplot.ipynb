## **Strip Plot**
```text
A strip plot shows individual observations for a categorical variable.

This is different from barplot() and countplot() because those summarize data into bars. stripplot() lets us see the actual data points.

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

### **Plotting our strip plot**

````python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.show()
````

```text
We may notice that several points overlap.
That's because multiple observations can have similar values.

### **jitter**
```text
jitter adds a small random horizontal displacement to the points.

````python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    jitter=True
)

plt.show()
````

```text
Jitter does not change the actual total_bill value. 
It only spreads points horizontally so overlapping observations become easier to see.
Its quite hard to see here because the data is quite vast.

````python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    jitter=False
)

plt.show()
````

```text
This how it behaves when jitter is False every value is in a single vertical line and might overlap if they have same observation.

### **hue**

````python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    jitter=True
)

plt.show()
````

### **dodge**
```text
Without dodge, the two groups can occupy the same categorical position.
With dodge=True, the groups are shifted side-by-side within each day.

````python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    dodge=True,
    jitter=True
)

plt.show()
````

### **size**
```text
Controls the size of each point.

````python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    jitter=True,
    size=3,
    hue="sex"
)

plt.show()
````

```text
below fig is same, we have just customised the legend to be on top left position of our figure.

````python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    jitter=True,
    size=3,
    hue="sex"
)
plt.legend(loc="upper left")
plt.show()
````

### **order**
```text
We have already seen how it works. It simply tells in which order we want our x-axis to be.
Like before it was Thur -> Fri -> Sat -> Sun

````python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    order=["Sun", "Sat", "Fri", "Thur"], # we have changed the previous order to this
    jitter=True
)

plt.show()
````

```text
now the order is Sun -> Sat -> Fri -> Thur 

## **Just another example combining all**

````python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    hue_order=["Female", "Male"],
    palette="Set2",
    jitter=True,
    dodge=True,
    size=5,
    alpha=0.7
)

plt.title("Total Bill Distribution by Day and Sex")
plt.xlabel("Day")
plt.ylabel("Total Bill")

plt.show()
````

| Parameter   | Purpose                    |
| ----------- | -------------------------- |
| `x`         | Categorical variable       |
| `y`         | Numerical variable         |
| `hue`       | Creates groups             |
| `jitter`    | Spreads overlapping points |
| `dodge`     | Separates hue groups       |
| `size`      | Point size                 |
| `alpha`     | Transparency               |
| `order`     | Category order             |
| `hue_order` | Hue-group order            |
| `palette`   | Group colors               |
