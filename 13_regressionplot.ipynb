## **Regression Plot**
```text
regplot() is used to show the relationship between two numerical variables and add a regression line that represents the overall trend.

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

### **Plotting our regression plot**

````python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
````

```text
Dots → Individual observations/data points.
Regression line → Shows the overall relationship/trend between the two variables.
Confidence interval → Shows the uncertainty around the estimated regression line.

### **scatter in regression plotting**

````python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    scatter=False
)

plt.show()
````

### **ci - the confidence interval**

````python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    ci=None
)

plt.show()
````
