## **Rug Plot**
```text
A rug plot is one of the simplest Seaborn plots. It shows the location of individual observations along an axis.

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

````python
tips["total_bill"].max()
````

````python
tips["total_bill"].min()
````

### **Plotting Rug Plot**

````python
sns.rugplot(
    data=tips,
    x="total_bill"
)

plt.show()
````

````python
sns.rugplot(
    data=tips,
    y="total_bill"
)

plt.show()
````

### *Important Parameters*
| Parameter | Purpose                         |
| --------- | ------------------------------- |
| `data`    | DataFrame                       |
| `x`       | Numerical variable along x-axis |
| `y`       | Numerical variable along y-axis |
| `height`  | Length of the rug marks         |
| `alpha`   | Transparency                    |


````python
sns.rugplot(
    data=tips,
    x="total_bill",
    height=0.05,
    alpha=0.5
)
````
