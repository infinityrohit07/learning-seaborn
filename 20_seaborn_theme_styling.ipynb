## **Seaborn Styling and Themes**


The main function is:
`sns.set_theme()`

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

### 1. **Default theme**

````python
sns.set_theme()

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
````

### 2. **style**

```text
Common styles:

"white"
"dark"
"whitegrid"
"darkgrid"
"ticks"

````python
sns.set_theme(style="whitegrid")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
````

### 3. **context**
```text
context controls the overall scale of plot elements, such as text and lines.

Common options:

"paper"
"notebook"
"talk"
"poster"

````python
sns.set_theme(
    style="whitegrid",
    context="paper"
)

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
````

```text
paper    → smaller
notebook → normal
talk     → larger
poster   → very large

This is useful when the same plot needs to be displayed at different sizes.

### **sns.set_palette()**
```text
Earlier, we used palette directly inside a plot:

sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    palette="Set2",
    errorbar=None
)

ut what if we want the same palette to be used by multiple plots?
That's where sns.set_palette() is useful.

````python
sns.set_palette("Set2")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex"
)

plt.show()
````

````python
sns.set_palette("Set2")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex"
)

plt.show()

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)

plt.show()
````

```text
You should use set_palette() when you're creating multiple plots and want consistent colors throughout your analysis.

### **sns.color_palette()**
```text
color_palette() is used to create or inspect a color palette.

1. **View a built-in palette**

````python
sns.color_palette("Set2")
````

2. **Create a palette with a specific number of colors**

````python
palette = sns.color_palette("Set2", n_colors=3)
palette
````

3. **Use the palette in a plot**

````python
palette = sns.color_palette("Set2", n_colors=2)

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    palette=palette
)

plt.show()
````
