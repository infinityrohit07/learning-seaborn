## **INTRODUCTION TO SEABORN**

### **WHAT IS SEABORN?**
```TEXT
Seaborn is a Python library used for data visualization and statistical visualization.

It is built on top of Matplotlib and provides high-level functions for creating informative and visually appealing plots with less code.

Key points:
Seaborn is a Python visualization library.
It is built on top of Matplotlib.
It works very well with Pandas DataFrames.
It provides convenient functions for statistical visualizations.
It provides better default styling than basic Matplotlib.
Matplotlib can still be used alongside Seaborn for additional customization.

### **Seaborn vs Matplotlib**
```TEXT
Both libraries can create many of the same types of plots, but they have different purposes and levels of abstraction.
```

| Matplotlib                                         | Seaborn                                      |
| -------------------------------------------------- | -------------------------------------------- |
| General-purpose visualization library              | Statistical/data visualization library       |
| Lower-level plotting interface                     | Higher-level plotting interface              |
| Provides detailed control                          | Makes common visualizations easier           |
| More manual customization may be required          | Good-looking defaults                        |
| Works with many types of data                      | Especially convenient with Pandas DataFrames |
| Foundation for many Python visualization libraries | Built on top of Matplotlib                   |


### **Why use Seaborn?**
```TEXT
Seaborn is particularly useful when working with datasets because it makes it easy to visualize relationships between variables and compare groups.

For example, you may want to answer questions such as:

Does studying more hours relate to higher marks?
How are salaries distributed?
How does salary differ between departments?
Are there outliers in the data?
Which variables are correlated?
How does a variable's distribution differ between categories?

Seaborn provides specialized plots for these types of questions.

### **Seaborn and Pandas**

````python
import pandas as pd

df = pd.DataFrame({
    "hours": [1, 2, 3, 4, 5],
    "marks": [35, 42, 50, 63, 72]
})

print(df)
````

````python
import seaborn as sns

sns.scatterplot(data=df, x="hours", y="marks")
````
