### **Violin Plot**
```text
A violin plot shows the distribution and shape of numerical data for each category.

Box plot information + KDE distribution shape

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

### **Plotting our Violin Plot**

````python
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.show()
````

```text
Wide → many observations around that value
Narrow → fewer observations around that value

### **hue & pallete**

````python
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    palette="colorblind"
)

plt.show()
````
