```text
pairplot() is used when you want to examine the relationships between multiple numerical variables at once.

```text
Instead of choosing just:

total_bill vs tip

you can give it the entire DataFrame:
```
`sns.pairplot(tips)`

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

## **Plotting our Pair Plot**

````python
sns.pairplot(tips)

plt.show()
````

### Understanding the `pairplot()`

The `pairplot()` creates a **grid of plots** to show the relationships between multiple numerical variables.

For our `tips` dataset, the numerical variables are:

- `total_bill`
- `tip`
- `size`

### How to read the grid

The grid has the same variables on both the **x-axis and y-axis**.

```text
                 total_bill       tip          size
              ┌─────────────┬─────────────┬─────────────┐
total_bill    │ distribution │  scatter    │  scatter    │
              ├─────────────┼─────────────┼─────────────┤
tip           │   scatter    │ distribution │  scatter    │
              ├─────────────┼─────────────┼─────────────┤
size          │   scatter    │  scatter    │ distribution │
              └─────────────┴─────────────┴─────────────┘

### **Customising our Pair Plot**

| Parameter   | Purpose                   |
| ----------- | ------------------------- |
| `vars`      | Select variables          |
| `hue`       | Separate data into groups |
| `palette`   | Control hue colors        |
| `kind`      | Change off-diagonal plots |
| `diag_kind` | Change diagonal plots     |


````python
sns.pairplot(
    tips,
    vars=["total_bill", "tip", "size"],
    hue="sex",
    palette="Set2",
    diag_kind="kde"
)

plt.show()
````
