## **Join Plot**
```text
jointplot() shows the relationship between two variables in the center and their individual distributions along the sides.

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

### **Plotting our Join Plot**

````python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
````

**1. Center → relationship**
```text
The center is a scatter plot:

X-axis → total_bill
Y-axis → tip
Each dot → one observation
```

**2. Top → total_bill distribution**
```text
The histogram at the top shows how the total_bill values are distributed.
```

**3. Right → tip distribution**
```text
The histogram on the right shows the distribution of tip.

### **kind parameter**
```text
The kind parameter controls what type of plot appears in the center.

| `kind`      | Center plot               |
| ----------- | ------------------------- |
| `"scatter"` | Scatter plot              |
| `"reg"`     | Scatter + regression line |
| `"kde"`     | KDE distribution          |
| `"hist"`    | 2D histogram              |
| `"hex"`     | Hexagonal bins            |


````python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="reg"
)

plt.show()
````
