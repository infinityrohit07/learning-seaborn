## **Count Plot**
```text
A countplot answers:
How many observations are there in each category?

Key differences:
countplot → counts rows
barplot   → calculates a statistic from a numerical column

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

````python
tips.head(10)
````

````python
tips.describe()
````

```text
this says the total number of tips that we have is 244.
but how many belongs to which day? we still don't know.

````python
tips.groupby("day")["total_bill"].count()
````

````python
tips.groupby("day").size()
````

### **Plotting our Counter Plot**

````python
bars=sns.countplot(
    data=tips,
    x="day"
)

plt.bar_label(bars.containers[0])
plt.show()
````

```text
See now we know that it exactly shows the number of rows for each day.

### **Customizing Count Plot**

````python
sns.countplot(
    data=tips,
    x="day",
    hue="sex",
    hue_order=["Female", "Male"],
    palette="Set2",
    order=["Thur", "Fri", "Sat", "Sun"],
    width=0.7,
    fill=True,
    edgecolor="black",
    linewidth=1,
    alpha=0.8
)

plt.show()
````

Note: WE CANNOT USE errorbar and estimators of bar plot here because they are bar plot specific
