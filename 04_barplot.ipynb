### **Bar Plot**
```text
A bar plot is used to compare a numerical value across different categories.

### **Importing seaborn and matplotlib and loading tips dataset**

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

````python
tips.head()
````

### **Plotting Bar Chart**

````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.show()
````

```text
In your plot:

Bar height → average total_bill for that day
Vertical lines in bars → uncertainty/error interval around that average

```text
We can remove them with the help of:
```
`errorbar=None`

````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    errorbar=None
)

plt.show()
````

````python
tips.groupby("day")["total_bill"].mean()
````

````python
bars = sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    errorbar=None
)

plt.bar_label(bars.containers[0])  # sns.barplot() returns a Matplotlib Axes object, not the bar container that bar_label() expects.

plt.show()
````

```text
This shows that by default, barplot() calculates the mean (average) for each category.

### **hue creates groups using color**

````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)

plt.show()
````

````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    errorbar=None
)

plt.show()
````

### **order**
```text
order lets us control the order in which categories appear on the x-axis.

````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    order=["Sun", "Sat", "Fri", "Thur"],
    errorbar=None,
    color="grey"
)

plt.show()
````

### **hue_order**
```text
While order controls the order of categories on the x-axis, hue_order controls the order of groups created by hue.

````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    hue_order=["Female", "Male"],
    errorbar=None
)

plt.show()
````

### **palette**
```text
It colors those groups.

````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    palette="Set2",
    errorbar=None
)

plt.show()
````

### **estimator**
```text
We already saw that by default, barplot() calculates the mean (average) for each category.

We can change which statistic is calculated using estimator.

````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    estimator="median", # this will now show median for each group
    errorbar=None
)

plt.show()
````

**Lets Verify**

````python
tips.groupby("day")["total_bill"].median()
````

````python
bars = sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    estimator="median",
    errorbar=None
)

plt.bar_label(bars.containers[0])

plt.show()
````

```text
estimator="sum"

````python
bars = sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    estimator="sum",
    errorbar=None
)

plt.bar_label(bars.containers[0], fmt="%.1f")
plt.show()
````

```text 
we can also create our own function and pass it as estimator

````python
def value_range(x):
    return x.max() - x.min() # for every group it will return the difference between the minimum and maximum value 

sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    estimator=value_range,
    errorbar=None
)

plt.show()
````

```text
We can also pass our numpy functions

````python
import numpy as np

sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    estimator=np.max, # This makes each bar represent the maximum bill for that day.
    errorbar=None
)

plt.show()
````

```text
The important concept is understanding that estimator accepts a function that reduces each group to one value.

### **fill**
```text
fill controls whether the bars are filled or not.

fill=True → filled bars
fill=False → outline-only bars

````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    fill=False,
    errorbar=None
)

plt.show()
````

````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    fill=True,
    errorbar=None
)

plt.show()
````

### **We can always customise them using our matplotlib skills**

````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    fill=True,
    errorbar=None,
    edgecolor="black",
    linewidth=2,
    width=0.6,
    alpha=0.8
)

plt.title("Average Total Bill by Day")
plt.ylabel("Total Bill")
plt.xlabel("Day")

plt.show()
````
